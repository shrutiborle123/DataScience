# args and kwargs

### 1.write positional argument then *args and then **kwargs

### Secquence-->1.positional args , 2.*args, 3. **kwargs

```python
def billling(**product):
    for key , value in product.items():
        print(key , value)

    amount = 0
    for k , v in product.items():
        amount += v
    return "total amount is :" , amount
```

```python
def mul(*nums):
    result = 1
    for i in nums:
        result *= i
    return result
```

```python
mul(10 , 20)
```
