# Lambda function

###we write lamda function in one line 

```python
add = lambda a , b : a + b
```

```python
add(2 , 4)
```

```python
greet = lambda : "Hello Worlds"
```

```python
greet()
```

```python
def greet():
    return "Hello World"

greet()
```

```python
num_check = lambda a : "even" if a % 2 == 0 else "odd"
```

```python
num_check(5)
```

```python
voting_card = lambda age : "Can apply" if age >=18 else "Can't apply"
```

```python
voting_card(19)
```

```python
voting_card(17)
```

```python
def small(word):
    word = word.lower()
    return word
```

```python
small("Hello, World!")
```

```python
small = lambda word : word.lower()
```

```python
small("HELLO")
```

```python
empty = lambda x : "string is empty" if x == "" else "not empty"
```

```python
empty("HII")
```

```python
num = lambda a: a**2
```

```python
num(3)
```
