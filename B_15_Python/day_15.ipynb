# Inhertance

```python
class vehicle:
    def __init__(self , brand):
        self.brand = brand
    def info(self):
        return f"Brand name is {self.brand}"
class car(vehicle):
    def __init__(self, brand, model):
        super().__init__(brand)
        self.model = model
    def car_info(self):
        return f"Brand name is {self.brand} and model is {self.model}"
```

```python
new_car = car("tata" ,"nexon")
new_car.info()
new_car.car_info()
```

```python
class Bank:
    def __init__(self , bank_name):
        self.bank_name = bank_name
    def info(self):
        return f"bank name is {self.bank_name}"
class user(Bank):
    def __init__(self, bank_name , user_name):
        super().__init__(bank_name)
        self.user_name = user_name
    def user_info(self):
        return f"User name is {self.user_name} and bank name is {self.bank_name}"
        
```

```python
bank_1 = Bank("SBI")
bank_1.info()
```

```python
user_1 = user("SBI" , "Shree")
user_1.user_info()
```
