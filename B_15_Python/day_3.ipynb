## operator

### Arithmatic operator


+ : addition : 5 + 3 = 8

- : substraction : 10 - 2 = 8

/ : division :7 / 2 = 3.5

// : floor division : 7 / 2 = 3

% : modulus : 7 / 2 = 1

* : multiplication : 2 * 3 = 5

** : exponentiation : 2**3 = 8 

```python
a = 5
b = 3
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a**b)
```
