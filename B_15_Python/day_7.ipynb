```python
for i in range(5):
    print(i)
```

```python
cart = ["bread" , "milk" , "eggs" , "oats"]
for i in cart:
    print("I'm purchasing", i)
```

```python
marks = []
for i in range(10):
    marks.append(100)
print(marks)
```

```python
speed = []
for i in range(5):
    speed.append(80)
print(speed)
```
