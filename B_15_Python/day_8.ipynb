## Pass, Break , Continue

```python
marks = [45,23,1,46,78,90,56,]
for i in marks:
    if i < 35:
        print("you have failed, need to improve your performance")
    print(i)  
```

```python
marks = [45,68,36,40,51,100,23] 
```

```python
for i in marks:
    if i >= 35:
        print("Pass",i)
    else:
        print("failed",i)
    break
```

```python
fruits = ["apple", "mango" , "cherry", "banana" ,"berry"]

for fruit in fruits:
    if fruit == "mango":
        print(fruit, "fruit found")
        break
```

```python
snacks = ["chips","taco","burger","ban maska" ,"oats", "softy"]
for i in snacks:
    if i == "oats":
        print("skip it please",i)
        continue
    print("yummy eating...!",i)
```

```python
marks = [89,34,45,78,93,23,0,70]
for i in marks:
    if i > 35:
        print("you've passed the exam....!",i)
    else:
        pass
```

```python
num = [1,2,3,4,5,6]
for i in num:
    if i == 4:
        print(i,"found the number")
    else:
        pass
```
