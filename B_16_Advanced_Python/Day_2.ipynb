# Slicing
### syntax - 1D array a[start_element, end_element(n-1)]
### 2D array: a[row_slicing, column_slicing]
### 3D array: a[block_slicing, row_slicing, column_slicing]

```python
#2D array
a = np.random.randint(1,10,(3,3))

print(a)
print(a[1:,1:])
print(a[0:,1:])
print(a[0:2,0:2])
print(a[1:,0:2])
print(a[0:1,0:])#1st row
print(a[0:,0:1])#1st column
```

```python
#3D array
ar = np.random.randint(1,10,(2,3,3))
print(ar)
print(ar[1:,1:,1:])
print(ar[0:1,0:2,1:])
print(ar[0:,0:,0:1])
print(ar[0:1,0:2,0:2])
print(ar[1:,0:2,0:2])
```
