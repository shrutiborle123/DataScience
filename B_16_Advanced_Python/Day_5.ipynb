```python
#merge function

d1 = pd.DataFrame({
    'ID':[101, 102, 103, 104, 105],
    'name':["A","B","C","D","E"]
})
d1
```

```python
d2 = pd.DataFrame({
    'ID':[101, 102, 103, 104, 105],
    'marks':[67, 45, 89, 100, 95]
})
d2
```

```python
data = pd.merge(d1,d2, on='ID')
data
```

```python
d3 = pd.DataFrame({
    'ID':[201, 202, 203, 204, 205],
    'name':["A","B","C","D","E"],
    'marks':[67, 45, 89, 100, 95]
})
d3
```

```python
dataset = pd.merge(d1,d3, on="name")
dataset
```
