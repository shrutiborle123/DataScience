# Matplotlib

### Charts
#### 1.line chart -use when we have to identify trend of overall time or overall period daywise, monthwise, yearwise
#### 2. bar chart -use to check categorical vs numerical column relation
#### 3. scatter plot -used only on numerical column, use to check relation between 2 numerical column
#### 4. histogram chart -use when we have distribution of data, plot only on 1 column, used only on numerical column
#### 5. pie chart - use when we have check distribution of in percentage
#### 6. Box plot -use to check outliers, also use to check 5 no. of summary
#### 7. violine chart -use for understanding data distribution and shape of data

### univarient - chart plot on single column --ex:histrogram
### Bivarient - chart plot on two columns -- all except histogram
### violine chart is used in both univarient, Bivarient

## line Chart

```python
mth = df.groupby('Month')['Openings'].sum()
mth
```

```python
plt.figure(figsize=(15,5)) # figure is used to adjust widht and height of plot first parameter is for width and 2ns is for height
mth.plot(kind="line", marker="o", color="red",linewidth=2, linestyle="--")
plt.title("Month wise number of openings")
plt.xlabel("Months")
plt.ylabel("Openings")
plt.xticks(rotation=0)
plt.grid()#add grid to the background of plot
plt.tight_layout()
plt.savefig("linechart.pdf")
plt.show()
```
