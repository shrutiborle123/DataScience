```python
data = pd.read_csv("india_job_market_2024_2026.csv")
```

```python
data.head(2)
```

## Bar Chart

```python
job = data.groupby("Job_Title")["Openings"].sum().sort_values(ascending=False).head(10)
job.plot(kind="bar", color = "purple", edgecolor = "black")
#job.plot(kind="barh", color = "purple", edgecolor = "black") barh for horizontal bar chart
plt.title("JOB OPENINGS")
plt.xlabel("job titles")
plt.ylabel("Openings")
plt.tight_layout()
plt.grid()
plt.xticks(rotation = 90)
plt.savefig("barchart.pdf")
plt.show()
```

## Scatterplot

```python
plt.scatter(data['Salary_LPA'], data["Applicants"], color = "red", edgecolor= "black", marker = "*", alpha=0.2)
plt.show()
```

## Histogram

```python
plt.hist(data["Company_Rating"], bins = 12, edgecolor = "navy", color = "violet") #bins are used no. of columns
plt.title("Company ratings")
plt.show()
```

```python
data["Date_Posted"] = pd.to_datetime(data["Date_Posted"])
data['year'] = data["Date_Posted"].dt.year
```

```python
year= data.groupby("year")["Openings"].sum()
year.plot(kind = "pie",autopct = "%1.2%%", shadow = True)
plt.show()
```
