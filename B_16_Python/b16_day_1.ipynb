```python
#assignment operator
x = 20
print("Value is:",x)

x += 5
print("After += value is:",x)

x -= 2
print("After -= value is:",x)

x *= 2
print("After *= value is:",x)

x /= 2
print("After /= value is:",x)

x %= 2
print("After %= value is:",x)

x **= 2
print("After **= value is:",x)

x //= 2
print("After //= value is:",x)


```
