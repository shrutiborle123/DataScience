```python
#3.Write a function that prints numbers from 1 to 10 using a function
def num():
    for i in range(1,11):
        print(i)
num()
```

```python
#4.Write a function that takes a number and returns whether it is even or odd.
def eve_odd():
    n = int(input("Enter The number:"))
    if n%2 == 0:
        print("Even")
    else:
        print("Odd")
eve_odd()
```

```python
#5.Write a function that takes two numbers and returns the greater number.
def Larger():
    a = int(input('Enter 1st Number:'))
    b = int(input('Enter 2nd number:'))
    if a > b:
        print("A is greater number",a)
    else:
        print("B is greater number",b)
Larger()
```

```python
#7.Write the Function add the parameter and check the number is positive, Negative or Zero.
def Num_Check(n):
    if n>0:
        print("positive")
    elif n<0:
        print("negative")
    else:
        print("it is Zero")
Num_Check(2)
Num_Check(-9)
Num_Check(0)
```

```python
#8.Write a function that prints the sum of numbers from 1 to 10.
def add():
    sum = 0
    for i in range(1,11):
        sum+=i
    print(sum)
add()
```

```python
#9.Write a function that takes a number and prints its table (multiplication table).
def table():
    n = int(input("Enter the number:"))
    for i in range(1,11):
        print(n,"*",i,"=",n*i)
table()
```

```python
#11.13. Create a function that returns the grade based on marks.
#* Marks ≥ 80 → A
#* Marks ≥ 60 → B
#* Marks ≥ 40 → C
#* Marks < 40 → Fail

def garde(g):
    if g >= 80:
        print("Your Grade is A:",g)
    elif g >= 60:
        print("Your Grade is B",g)
    elif g >= 40:
        print("Your Grade is C",g)
    else:
        print("Fail",g)
garde(89)
garde(45)
garde(12)
```
