## Quetions Of Args and Kwargs

```python
#1.Write a function using *args that prints all numbers passed to it.
def num(*args):
    print(args)
num(2, 23, 43, 6, 4, 78)
```

```python
#2.Write a function using *args to find the sum of all numbers.
def add(*args):
    print("Sum of numbers is:",sum(args))
add(5,5,4,2)
```

```python
#3. Write a function using *args to count how many values are passex
def count_val(*args):
    c = 0
    for i in args:
        c+=1
        print(i, end=" ")
    print("Total values passed:",c)
count_val(1,2,3,4)
```

```python
#3. Write a function using *args to count how many values are pass
def count_val(*argus):
    print(len(argus))
count_val(2,6,45,3,9)
```

```python
#4.Write a function using *args that multiplies all numbers together.
def multiplication(*args):
    mul = 1
    for i in args:
       print(i, end=" ")
       mul = mul*i
    print("\nMultiplication of all numbers is:",mul)
multiplication(2,3,4)
```

```python
#5.Write a function using **kwargs that prints the dictionary it receives.
def dict(**kwargs):
    print(kwargs)
dict(name = "shree", age = 22, city = "pune")
```

```python
#6.Write a function using **kwargs to print only the keys.
def dict(**kwargs):
    print(kwargs.keys())
dict(name = "shree", age = 22, city = "pune")
```

```python
#7.Write a function using **kwargs to print only the values.
def dict(**kwargs):
    print(kwargs.values())
dict(name = "shree", age = 22, city = "pune")
```

```python
#8.Write a function using **kwargs that counts how many key–value pairs are passed.
def dict(**kwargs):
    print(len(kwargs))
dict(name = "shree", age = 22, city = "pune")
```

```python
def dict(**kwargs):
    c = 0
    for k,v in kwargs.items():
        c +=1
    print(c)
    
dict(name = "shree", age = 22, city = "pune")
```

# Lambda Function

```python
#Write a lambda function to add two numbers.
a = lambda x,y: x+y
print(a(5,5))
```

```python
#2. Write a lambda function to square a number.
sq = lambda s: s**2
print(sq(3))
```

```python
#3. Write a lambda function to find the cube of a number.
cube = lambda x: x**3
print(cube(3))
```

```python
#4. Write a lambda function to check if a number is even.

eve_odd = lambda x: "even" if x%2 == 0 else "odd"
print(eve_odd(4))
```

```python
#5. Write a lambda function to find the Largest Numbers
large_num = lambda a,b : "a is largest" if a> b else "b is largest"
print(large_num(5,6))
```

```python
#5. Write a lambda function to find the Largest Numbers
l = lambda a,b: a if a>b else b
print("The largest number is :",l(5,6))
```

# Recursive Function

```python
def info(n):
    if n == 0:
        print("Done")
    else:
        print(n)
        info(n-1)
info(5)
```

```python
def fact(n):
    if n == 1: #base case
        return 1
    else:
        return n*fact(n-1) #recursive call
fact(5)
```
