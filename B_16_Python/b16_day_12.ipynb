## List Comprehension

#### - when output is needed in list then use list comprehension
#### - Syntax: var = [expression for loop if condition]

```python
c = [i for i in range(1,11)]
print(c)
```

```python
eve = [i for i in range(1,11) if i%2 == 0]
print(eve)
```

## Questions Of list comprehension

```python
1. Create a list of numbers from 1 to 10 using list comprehension.
2. Create a list of squares of numbers from 1 to 10.
3. Create a list of even numbers from 1 to 20.
4. Given a list numbers = [1, 2, 3, 4, 5], create a new list where each number is multiplied by 5.
5. Given a string "python", create a list of its characters using list comprehension.
6. Given a list numbers = [10, 15, 20, 25, 30], create a list of numbers divisible by 5.
7. Create a list of numbers from 1 to 50 that are divisible by both 3 and 5.
8. Given a list names = ["om", "rahul", "priya"], create a new list where each name is converted to uppercase.
9. Given a list words = ["apple", "bat", "cat", "elephant"], create a list containing the length of each word.
10. From a list of marks [45, 67, 89, 34, 90], create a list of "Pass" or "Fail" (pass ≥ 50)
```

```python
#1. Create a list of numbers from 1 to 10 using list comprehension.
c = [i for i in range(1,11)]
print(c)
```

```python
#2.Create a list of squares of numbers from 1 to 10.
sq = [i**2 for i in range(1,11)]
print(sq)
```

```python
#3.Create a list of even numbers from 1 to 20.
even = [i for i in range(1,21) if i%2 == 0]
print(even)
```

```python
#4.Given a list numbers = [1, 2, 3, 4, 5], create a new list where each number is multiplied by 5.
multiply = [numbers*5 for numbers in range(1,6)]
print(multiply)
```

```python
#5.Given a string "python", create a list of its characters using list comprehension.
c = [ a for a in "python"]
print(c)
```

```python
#6. Given a list numbers = [10, 15, 20, 25, 30], create a list of numbers divisible by 5.
l = [10, 15, 20, 25, 30]
d = [i for i in l if i%5 == 0]
print(d)
```

```python
#7.Create a list of numbers from 1 to 50 that are divisible by both 3 and 5.
div = [i for i in range(1,51) if i%3 == 0 and i%5 == 0]
print(div)
```

```python
#8.Given a list names = ["om", "rahul", "priya"], create a new list where each name is converted to uppercase.
l = ["om", "rahul", "priya"]
u = [i.upper() for i in l]
print(u)
```

```python
#9.Given a list words = ["apple", "bat", "cat", "elephant"], create a list containing the length of each word.
l = ["apple", "bat","cat","elephant"]
c = [len(i) for i in l]
print(c)
```

```python
#10.From a list of marks [45, 67, 89, 34, 90], create a list of "Pass" or "Fail" (pass ≥ 50)
marks = [45,67,89,34,90]
pf = ["pass" if i>=50 else "fail" for i in marks]
print(pf)
print(marks)
print(list(zip(marks,pf)))
```

## DateTime Module

```python
import datetime as d #datetime is module
import pytz # module for timezone

z = pytz.timezone('Asia/Kolkata')

da = d.datetime.now(z) #now function is use to print current date and time both
print(da) 
```

```python
import datetime as d #datetime is module
import pytz # module for timezone

z = pytz.timezone('Asia/Kolkata')

da = d.datetime.now(z).time() #time function is use to print current time
print(da) 
```

```python
import datetime as d #datetime is module
dt = d.date.today() #today is use to print current date
print(dt)
```

```python
import datetime as d #datetime is module

D = d.datetime(2025,10,14,4,12,13) #we can Enter date and time Manuaaly by using this syntax
print(D)
```
