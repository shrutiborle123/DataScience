## (Object Oriented Programming)

```python
class Students:
    name = "aditya kale"
obj = Students()
print(obj.name)
```

```python
class Teacher:
    def show(self):
        print("This is my first class")
        print(self == obj)
        print(self is obj)
obj = Teacher()
obj.show()
```

```python
class Student:
    
    def info(self, name, age, dept):
        print("The name is:",name)
        print("The age is:",age)
        print("The Department is:",dept)
    
    def show(self,city):
        #print("The name is:",name) 
        print("the city is:",city)

s1 = Student()
s1.info("Shree" , 22, "DS and Ml")
s1.show("Pune")
```

```python
class Car:
    def display(self, name, price):
        self.name = name
        self.price = price
        print("The Name is:",name)
        print("The Price is:",price)
        
    def show(self, color):
        print("The Name is:",self.name)
        print("The Price is:",self.price)
        print("The color is:",color)

c1 = Car()
c1.display("Mahindra", 1500000)
c1.show("White")
```

```python
# Constructor

class Details:
    def __init__(self):
        print("The name is shree")
        print("The age is 22")
        print("The City is mumbai")
d1 = Details()
```

```python
# Constructor

class Details:
    def __init__(self, n, a, c):
        print("The name is:",n)
        print("The age is:",a)
        print("The City is:",c)
d1 = Details("Shree", 22, "Mumbai") # arguments of parameters of constructor is passed directly in object of class
```
