```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def display(self):
        print("The name of the Student is:",self.name)
        print("The age of the student is:",self.age)
        
s1 = Student("Shruti",22)
s1.display()
        
        
        
      
```

```python
class Car:
    
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color
    
    def show(self):
        print("The Brand Of car is:",self.brand)
        print("The color Of car is:",self.color)

c1 = Car("Tata" ,"Black")
c2 = Car("Mahindra" ,"Brown")
c1.show()
c2.show()
```

```python
class Rectangle:
    
    def __init__(self, l, b):
        self.l = l
        self.b = b
        
    def display(self):
        print("Area of rectangle is:",self.l*self.b)
        
r1 = Rectangle(2, 4)
r1.display()
        
```

```python
class Mobile:
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price
        
    def show(self):
        p = self.price*10/100
        print("The Brand is:",self.brand)
        print("The Price is:",self.price)
        print("The discount amount is:",p)
        print("The price after Discount of 10% is:",self.price-p)
m1 = Mobile("MoTo", 30000)
m1.show()
        
```

```python
class Employee:
    
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
        
    def info(self):
        s = self.salary*10/100
        print("The name is:",self.name)
        print("The Salary is:",self.salary)
        print("the salary after increament of 10% is:",self.salary+s)

e1 = Employee("Shree", 50000)
e1.info()
        
    
        
```

```python
class Circle:
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        a = 3.14*(self.radius**2)
        print("The area of circle is:",a)
c = Circle(4)
c.area()
    
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
    def elegible(self):
        if self.age >=18:
            print("Elegible For Voting.",self.age)
        else:
            print("Not Eligiblr For Voting.",self.age)
obj = Person(19)
obj.elegible()
    
```

# Inhertance


```python
#single Inhertance
class Parent:
    def teach(self):
        print("this is parent class.")
    
class child(Parent):
    def show(self):
        print("This is child class.")
c3 = child()
c3.show()
c3.teach()
    
```
