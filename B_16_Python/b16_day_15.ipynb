```python
#multiple Inhertance
class Father:
    def A(self):
        print("This is Father Class")
class Mother:
    def B(self):
        print("This is Mother Class")
class Child(Father,Mother):
    def C(Self):
        print("This is child class")
obj = Child()
obj.A()
obj.B()
obj.C()        
        
```

```python
#Multilevel Inhertance
class GrandFather:
    def A(self):
        print("This is Grandfather Class")
        
class Father(GrandFather):
    def B(self):
        print("This is Father class")
        
class Child(Father):
    def C(self):
        print("This is cHild  Class")
        
class Son(Child):
    def D(self):
        print("This Is Son Class")
        
s1 = Son()
s1.A()
s1.B()
s1.C()
s1.D()
```

```python
# Heirarchical Inhertance

class Parent:
    def data(self):
        print("This is parent class")
class Child1(Parent):
    def X(self):
        print("This is Child1 Class")
class Child2(Parent):
    def Y(self):
        print("This is Child2 Class")

c1 = Child1()
c2 = Child2()

c1.data()
c1.X()
 
c2.data()
c2.Y()
```

```python
# Hybrid Inheritance
class School:
    def info(self):
        print("This is school ")
class Student(School):
    def data(self):
        print("This is Student")
class Teacher(School):
    def tech(self):
        print("This is teacher")
class assistant(Student,Teacher):
    def guide(self):
        print("Help")
        
        
a1 = assistant()
a1.info()
a1.data()
a1.tech()
a1.guide()
```

# PolyMorphism

```python
#method OverLoading
class Dog:
    def sound(self):
        print("Bow Bow........!")
        
class Cat:
    def sound(self):
        print("Meow Meow.......!")

d = Dog()
c = Cat()
d.sound()
c.sound()
```

```python
# Method overriding
class Dog:
    def sound(self):
        print("Bow Bow........!")
        
class Cat(Dog):
    def sound(self):
        print("Meow Meow.......!")

c = Cat()
c.sound()
```

```python
#Method OverLoading
class Math:
    def Calc(self,l, w=None): # we can also *arg Here
        if w is None:
            print("Square is:",l**2)
            
        else:
            print("Area Of Rectangle is:",l*w)
M1 = Math()
M1.Calc(5) # One method Performing Multiple Operation
M1.Calc(5,10)
```
