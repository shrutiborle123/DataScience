# Encapsulation

```python
class Practice:
    def __init__(self):
        self.name = "Ravan" #public
        self._age = 100 #protected we use "_""single underscore for protected"
        self.__city = "Lanka" #private we use "__""double underscore for private"
        
    def show(self):
        print(self.__city)
obj = Practice()
print(obj.name)
print(obj._age)
obj.show()
```

```python
class Bank:
    def __init__(self, name, age, bal):
        self.name = name
        self._age = age
        self.__bal = bal
    def deposite(self, amt):
        self.__bal += amt
        print("deposite amount is:",amt)
        print("Total amount After Deposite is:",self.__bal)
    def withdraw(self, AMT):
        if self.__bal >= AMT:
            self.__bal -= AMT
            print("Withdrawn Amount is:",AMT)
            print("Remaining Balance is:",self.__bal)
        else:
            print("Sorry.....! Insufficiant Balance.")
OBJ = Bank("Khushi", 22, 25000)
OBJ.deposite(5000)
OBJ.withdraw(5000)
```

# Abstraction

```python
from abc import ABC,abstractmethod

class Payment(ABC):  
    @abstractmethod
    def Pay(self): #if we create object of this method the data will not print 
                   #and it'll give error bcoz we abstracted this method
        pass
    
class UPI(Payment):
    def Pay(self):
        print("Money Paid Using UPI....")
        
class CreditCard(Payment):
    def Pay(self):
        print("Money paid using Credit Card.....")

OBJ1 = UPI()
OBJ2 = CreditCard()

OBJ1.Pay()
OBJ2.Pay()
```
