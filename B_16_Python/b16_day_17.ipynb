# Exception Handling

```python
#Zerodevision error.
try:
    a = int(input("Enter the number:"))
    print(5/a)
except: # it'll get run when code has error
    print("Enter Valid number.")
else: # it'll get executed when the code is correct and error free
    print("Code Executed Successfully.")
finally:# it'll get run wheather their is error is not
    print("All Execution is Done.")
```

# Regular Expression

```python
import re

txt = "Hello...!, i'm Shruti Borle. My mobile number is 1234567899. My age is 22"

print(re.match(r"Hello",txt))

print(re.search(r"Mobile",txt))

print(re.sub(r"Shruti","Shree",txt))

print(re.findall(r"\d+",txt))

print(re.split(r"\s+",txt))
```
