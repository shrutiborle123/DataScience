# List

## Methods in List

```python
list = [10 , 20 , "shree" , "ram" , "Pune" , True , 58.65 , "Mumbai" , 740000]
print(list)
print(len(list))
```

```python
list[4] = "nagpur"
print(list)
```

```python
#indexing 
print(list[4])

print(list[1])

print(list[-4]) #nagative indexing

print(list[-1])

```

```python
#slicing
print(list[4:8]) # it print the element till the second last index of end 

print(list[-4:-1])

print(list[4:]) # if we did'nt pass end then it will print till the last element

#syntax of slicing[start : end : steps] -step is by default 1
```

```python
#reverse list or string with ::-1
'''[::-1] is Python slicing syntax: [start:stop:step].

The first two parts are omitted, so Python uses the whole string or list.
-1 is the step, meaning move backward one item at a time.
'''
print(list[::-1])
```

```python
print(list)
```

```python
list.append("Ravan") # add element to the end of the list
print(list)
```

```python
l = [10 , 2 , 5 , 7 , 3 , 9]
print(l, type(l))
```

```python
l.sort() #ascending 
print(l)
```

```python
l.sort(reverse = True) #decending
print(l)
```

```python
l.insert(2, "Shree") #insert element at the index you given 
                     #--- syntax insert(index , item/ element)
print(l)
```

```python
l.extend(["arjun" , "Aditya" , "Karan"]) # add multiple element to the list

print(l)
```

```python
l.remove("Shree") #rempve iten by element
                 # remove first occurance of element
print(l)
```

```python
l.pop(5) #remove element by index
print(l)
```

```python
l.clear() #remove all the element of the list and return empty list
print(l)
```
