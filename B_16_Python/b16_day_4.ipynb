# Tuple

### - tuple is built in function or data types where  we store diffrent types of data in one variable
### - prenthesis () is used to store data.
### - immutable in python 
### - ordered sequence of data 
### - we use indexing and slicing in tuple
### - tuple is faster than list  

## Methods
### 1.Count()
### 2.Index()

```python
a = (10, 20, 30, 40, 50,60)
print(a, type(a))
```

```python
a = (10) #integer , tuple does not store single value
print(a,type(a))

s = (10,)
print(s,type(s))
```

```python
t = (100 , "Shree", True, 34.12, 22)
print(t, type(t))
print(len(t))
```

```python
t[0] = 12 #tuple is immutable 
print(t)
```

```python
import timeit
list = timeit.timeit("[10, 20, 30, 40, 50]")
tup = timeit.timeit("(10, 20, 30, 40, 50)")

print("The time of execution of list is:",round(list,2))
print("The time of execution of tuple is:",round(tup,2))
```

# Set

### -{} is used to store or define set
### -Does not allow duplicate values
### - stores only unique values
### - mutable in python 
### - unordered sequence
###  *x={} is dictionary by default ---we cant define empty set

```python
s = {10, 20, 30, 40,}
print(s,type(s))
```

```python
s = {}
print(s,type(s)) #by default dictionary, we cant define empty set

a = {10,}
print(a, type(a))
```

```python
s = {100, 43, 100, 32, 67, 100, 800, 56, 70}
                  #does not store duplicate values
                  #and unordered sequence
print(s)
```

```python
s.add("Shree")
print(s)
```

```python
s.remove(67)
print(s)
```

```python
s.pop() #remove first element
print(s)
```

```python
s.update(["Shruti", "Khushi", "Vaishu"])
print(s)
```

```python
s.discard("Shruti")
print(s)
```

```python
s.clear()
print(s)
```

```python
a = s.copy()
print(a)
```

```python
s1 = {1, 3, 5, 3, 4,}
s2 = {1, 2, 7, 8, 9}
print(s1.union(s2)) # combine both sets and remove duplicate
print(s1.intersection(s2)) # same element from set
print(s1.difference(s2)) # return uniqe element
```
