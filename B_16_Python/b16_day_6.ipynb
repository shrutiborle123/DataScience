# Dictionary
### -Store data in key value pair
### - {Kay:Value}
### - Mutable
### - key can be unique
### - Values can be duplicate
### - Methods : 1.keys(), 2.get(), 3. values(), 4.items(), 5.pop(), 6. clear(), 7.update()

```python
d = {
    "name" : "Shree",
    "age" : 22,
    "city" : "pune",
    "salary" : 780000
}
print(d)
```

```python
print(d["age"])
```

```python
d["age"] = 25
print(d)
```

### pop is a method and del is built in function use to remove key value pair

```python
print(d.keys())
```

```python
print(d.values())
```

```python
print(d.get("age"))
```

```python
print(d.items())
```

```python
print(d.pop("age"))
```

```python
print(d)
```

```python
del d["city"]
```

```python
print(d)
```

```python
d.update(
    {
        "name":"Shree",
        "city" : "pune"
    }
)
```

```python
print(d)
```

```python
print(d.clear())
```

# Conditional Statement

## 1.if-else:

```python
n = int(input("Enter The Number:"))

if n > 0:
    print("Positive")
else:
    print("Negative")
```

```python
marks = int(input("Enter Marks:"))
if marks >= 35:
    print("Pass")
else:
    print("Fail")
```

```python
n = int(input("Enter The NUmber:"))
if n % 2 == 0:
    print("Even")
else:
    print("Odd")
```

## 2.if-elif-else:

```python
a = int(input("Enter 1st Number:"))
b = int(input("Enter 2nd number:"))
op = input('Enter Operator (+,-,*,/):')

if op == "+":
    print("Sum:", a+b)
elif op == "-":
    print("Diffrence:", a-b)
elif op == "*":
    print("Multiplication:", a*b)
elif op == "/":
    print("Division:" , a/b)
else:
    print("Please Enter valid Operator")
```
