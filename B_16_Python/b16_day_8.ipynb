### - Break - break loop and iteration
### - continue - Skip 1 iteration and continue loop

```python
for i in range(1,11):
    if i == 6:
        break
    print(i)
```

```python
for i in range(1,11):
    if i == 6:
        continue
    print(i, end=" ")
```
