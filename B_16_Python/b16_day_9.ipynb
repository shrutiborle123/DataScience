## While Loop

```python
i = 1
while (i <= 10):
    print(i)
    i+=1
```

```python
i = 10
while(i>=1):
    print(i , end = " ")
    i-=1
```

```python
i = 2
while(i <= 20):
    if i%2 == 0:
        print(i)
    i+=1
```

```python
i = 1
while(i <= 20):
    if i%2 != 0:
        print(i)
    i+=1
```

```python
sum = 0
i=1
while(i<=100):
    sum+=i
    i+=1
print(sum)
```

```python
i = 1
n = 2
while(i <= 10):
    print("2 *", i,"=", n*i)
    i+=1
```

```python
n = int(input("Enter Number:"))
i = 1
while(i <= 10):
    print(n, "*" ,i, "=" ,n*i)
    i+=1
```

```python
i = 1
while(i <= 10):
    print(i**2)
    i+=1
```

```python
l = []
i = 1
while(i <= 20):
    l.append(i)
    i+=1
print(l)
```

```python
a = "python"
c = 0
i = 1
while(i <= len(a)):
    c+=1
    i+=1
print(c)
```
